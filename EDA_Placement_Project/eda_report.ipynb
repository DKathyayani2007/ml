#Student Placement Data Analysis

##Justification

The Student Placement Dataset was selected for this project because it contains important academic and placement-related information of students. This dataset is suitable for Exploratory Data Analysis (EDA) as it helps in understanding the relationship between academic performance, work experience, specialization, and placement status. The dataset also contains missing values and different types of variables, making it useful for data cleaning, preprocessing, statistical analysis, and visualization. Therefore, it is an appropriate dataset for performing EDA and generating meaningful insights.

```python
import pandas as pd

df = pd.read_csv("Placement_Data_Full_Class copy.csv")
print(df.head())
print(df.columns)
```

```python
# Fill missing values in salary column
df['salary'] = df['salary'].fillna(df['salary'].median())

# Remove duplicate rows
df = df.drop_duplicates()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.to_csv("cleaned_dataset.csv", index=False)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x='status', data=df)
plt.show()
```

```python
sns.countplot(x='workex', hue='status', data=df)
plt.show()
```

```python
sns.boxplot(data=df)
plt.show()
```

```python
sns.scatterplot(x='ssc_p', y='salary', data=df)
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True)
plt.show()
```

#Insights for ML Modeling

##Outliers

The box plot analysis shows that some salary values are significantly higher than the majority of observations. These outliers may affect certain machine learning algorithms such as Linear Regression and should be examined before model training.

##Collinearity

The correlation heatmap indicates the relationships among numerical features such as SSC percentage, HSC percentage, Degree percentage, Employability Test percentage, and MBA percentage. Highly correlated features can lead to multicollinearity, which may affect the performance and interpretability of some machine learning models.

##Feature Selection Considerations

Academic performance indicators and work experience appear to be relevant features for predicting placement status. These features can be considered during feature selection and model development.

##Data Quality

Missing values in the salary column were handled using the median, and duplicate records were removed. The dataset is now cleaned and ready for further machine learning analysis.
